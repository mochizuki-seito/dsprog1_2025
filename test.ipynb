# HTTPリクエスト

## 基本のリクエスト

```python
# アクセスしたいwebサイトのURLを指定
url='https://www.musashino-u.ac.jp/'

# webサーバーにHTTPリクエストを送信
# レスポンスを変数に格納しておく
res = requests.get(url)
```

```python
print(f"レスポンス: {res}") # レスポンスオブジェクトを表示
print(f"レスポンスの型: {type(res)}") # レスポンスオブジェクトの型を表示

print(f"リクエスト: {res.request}") # リクエストオブジェクトを表示
print(f"リクエストの型: {type(res.request)}") # リクエストオブジェクトの型を表示
print(f"リクエストヘッダー: {res.request.headers}") # リクエストヘッダーを表示
print(f"リクエストメソッド: {res.request.method}") # リクエストメソッドを表示
print(f"リクエストURL: {res.request.url}") # リクエストURLを表示
```

```python
# レスポンス
res.encoding = res.apparent_encoding  # レスポンスのエンコーディングを設定

print(f"レスポンスヘッダー: {res.headers}") # レスポンスヘッダーを表示
print(f"レスポンスボディ: {res.text}") # レスポンスボディを表示
```

## BeautifulSoupでHTMLソースからの任意の情報を取得する

```python
url = "https://www.musashino-u.ac.jp/"
res = requests.get(url, headers=headers)
res.encoding = res.apparent_encoding  # レスポンスのエンコーディング
print(f"ステータスコード: {res.status_code}") # ステータスコードを表示
```

```python
# HTMLソースコードを解析
soup = BeautifulSoup(res.text, 'html.parser')
print(f"soupの型: {type(soup)}")  # BeautifulSoupオブジェクトの型を表示
```

```python
# ソースコード全体から<a>タグを抽出
soup_a = soup.a
print(f"<a>タグ : {soup_a}")  # <a>タグを表示
print(f"<a>タグの型: {type(soup_a)}")  # <a>タグの型を表示

# ソースコード
print(f"<a>タグの中の<a>タグ: {soup_a.a}")
# 最初に見つかった<a>タグを表示
print(f"<a>タグの中の<a>タグの型: {type(soup_a.a)}") # <a>タグオブジェクトの型を表示
```

```python
# ソースコード全体から<div>タグを抽出
soup_div = soup.div.div
# print(f"<div>タグ: {soup_div}")
# 最初に見つかった<div>タグを表示
# print(f"<div>タグの型: {type(soup_div)}") # <div>タグオブジェクトの型を表示
```

```python
# BeautifulSoupオブジェクトとTagオブジェクトは,"."でHTMLタグを連結することでより深くアクセスする
soup_div_div = soup_div.div
print(f"<div>タグの中の<div>タグ: {soup_div_div}")
# 最初に見つかった<div>タグを表示
print(f"<div>タグの中の<div>タグの型: {type(soup_div_div)}") # <div>タグオブジェクトの型を表示
```

```python
# Tagオブジェクト.stringでタグ内のテキスト情報を取得する
print(f"<a>タグのテキスト: {soup_a.string}")  # <a>タグ内のテキストを表示

# コンテンツがない場合はNoneが返される
print(f"<div>タグの中の<div>タグのテキスト: {soup_div.string}")  #
```

## 指定してHTMLタグで要素を検索

```python
# 指定したタグが使われている要素のうち、最初に見つかった要素を取得する
h2 = soup.find('h2')
print(f"{h2}")  # 最初に見つかった<h2>タグを表示
print(f"<h2>変数の型: {type(h2)}")  # <h2>タグオブジェクトの型を表示
```

```python
h2_all = soup.find_all('h2')
# print(f"<h2>タグのリスト: {h2_all}")  # 見つかった<h2>のタグ
# print(f"型: {type(h2_all)}")  # <h2>型を表示

for h2 in h2_all:
    if h2.span is not None:
        print(f"{h2.span.string}")  # 各<h2>タグ内のテキストを表示  
```

```python
# a_list = soup.find_all('a')
# for a in a_list:
#    print(f"{a.string}")  # 各<a>タグ内のテキストを表示

# 検索結果の数に制限をかける
a_list_limited = soup.find_all('a', limit=5)
for a in a_list_limited:
    print(f"{a.string}") # 各<a>タグ内のテキストを表示
```

```python
# 複数のタグを指定して検索することも可能
# 例えば、<h2>タグと<h3>タグを両方を検索する場合は以下のようにする
h2_h3_list = soup.find_all(['h2', 'h3'])
for tag in h2_h3_list:
    print(f"{tag}")  # <h2>と<h3>タグを表示
```

## 任意の属性で検索

```python
tag_href_list = soup.find_all(attrs={'href': "/"}, limit=5)
for tag in tag_href_list:
    print(f"{tag}")  # href属性が"/"のタグを表示
```

```python
# id属性やclass属性など、任意の属性で検索することも可能である
# 例えば、id属性が"main-content"のタグを検索する場合は以下のようにする
tag_id_list = soup.find_all(id="top")
for tag in tag_id_list:
    print(f"{tag}")  # id属性が"top"のタグを表示


tag_class_list = soup.find_all(class_="p-notices", limit=5)
for tag in tag_class_list:
    print(f"{tag}")  # class属性が"p-notices"のタグを表示
```

# 練習問題
下記のページから出演タイトルを取得する
https://abehiroshi.la.coocan.jp/movie/eiga.htm

```python
url = "https://abehiroshi.la.coocan.jp/movie/eiga.htm"
res = requests.get(url)
res.encoding = 'Shift_JIS'
soup = BeautifulSoup(res.text, 'html.parser')

# ターゲットのテーブルを取得（border="1" のテーブルを想定）
target_table = soup.find('table', attrs={'border': '1'})
if target_table is None:
    # フォールバック: テーブルが複数ある場合は最後のテーブルを使う
    tables = soup.find_all('table')
    if tables:
        target_table = tables[-1]

if target_table is None:
    raise RuntimeError("ターゲットのテーブルが見つかりませんでした。HTML構造を確認してください。")

# タイトル一覧を格納するリスト
movie_titles = []

# テーブル内の行を処理
for row in target_table.find_all('tr')[1:]:  # ヘッダー行をスキップ
    cells = row.find_all('td')
    if len(cells) >= 2:
        # タイトルセルからテキストを抽出
        title_text = cells[1].get_text().strip()
        # 不要な全角スペース等を正規化
        title_text = title_text.replace('\u3000', ' ').strip()
        movie_titles.append(title_text)

# 結果を表示
print("阿部寛の出演映画タイトル一覧:")
for title in movie_titles:
    print(title)

print(f"\n合計: {len(movie_titles)}作品")
```
